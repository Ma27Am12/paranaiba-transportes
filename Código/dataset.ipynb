# Criando dataset da empresa

```python
import pandas as pd 
import numpy as np
```

Criando identificador dos caminhões.

```python

ids = [f"10{str(i).zfill(2)}" for i in range(1, 101)]

veiculo = pd.DataFrame({"Número_caminhao": ids})

```

```python
veiculo.head()
```

```python
veiculo.tail()
```

Verificando o tipo de dado da coluna

```python
veiculo.info()
```

```python
marcas = ['Scania', 'Mercedes', 'Volvo']
veiculo['marca'] = np.random.choice(marcas, size=len(veiculo))
```

```python
veiculo.head()
```

```python
veiculo.tail()
```

```python
start = pd.to_datetime("2012-01-01")
end = pd.to_datetime("2023-12-31")

n = len(veiculo)

random_dates = start + (end - start) * np.random.rand(n)

veiculo['Ano'] = random_dates
```

```python
veiculo.head()
```

```python
veiculo.tail()
```

```python
ano_de_fabricação = pd.to_datetime(veiculo['Ano']).dt.year
```

```python
ano_de_fabricação
```

```python
km = []
for ano in ano_de_fabricação:
    if ano <= 2014:
        km.append(np.random.randint(600000, 900000))
    elif ano <= 2018:
        km.append(np.random.randint(300000, 600000))
    else:
        km.append(np.random.randint(50000, 300000)) 
veiculo['Km_rodados'] = km
```

```python
veiculo.head()
```

```python
tipo = ['rígido', 'articulado']

veiculo['tipo'] = np.random.choice(tipo, size=len(veiculo))
```
