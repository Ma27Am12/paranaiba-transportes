```python
produtos_list = [
    'glp', 
    'Disel', 
    'Gasolina', 
    'soja',
    'milho',
    'café',
    'eletrônicos',
    'Carnes',
    'Queijo'
]
```

```python
n_entregas = 1000
```

```python
entregas = pd.DataFrame({"Produtos": np.random.choice(produtos_list, size=n_entregas)})
```

```python
entregas['id_entrega'] = range(1, n_entregas + 1)

cols = ['id_entrega'] + [col for col in entregas.columns if col != 'id_entrega']
entregas = entregas[cols]
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
start = pd.to_datetime('2020-01-01')
end = pd.to_datetime('2025-12-31')

random_entregas_dates = start + (end - start) * np.random.rand(n_entregas) 

entregas['Data_da_entrega'] = pd.to_datetime(random_entregas_dates).normalize()
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
status_lista = ['A tempo', 'Atrasado']
entregas['Status'] = np.random.choice(status_lista, size = len(entregas))
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
entregas.head()
```

```python
entregas['Destino'] = np.random.choice(rotas_lista, size = len(entregas))
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
entregas['Distância_km'] = entregas['Destino'].map(distancias)
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
consumo_medio = 3.5
preco_disel = 6.20

entregas['Custo'] = (entregas['Distância_km'] / consumo_medio) * preco_disel
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
preco_por_km = 5.0
entregas["Faturamento"] = np.where(
    entregas["Status"] == "Atrasado",
    entregas["Distância_km"] * preco_por_km * 0.90,  
    entregas["Distância_km"] * preco_por_km          
)
```

```python
entregas.head()
```

```python
entregas.tail()
```

```python
entregas.to_csv('C:/Users/mathe/Documents/projetos/Analise de dados/projeto-20/paranaiba-transportes/dataset/entregas.csv')
```
